# Corrélations des données

Dans ce notebook, nous chercherons des [corrélations](https://fr.wikipedia.org/wiki/Corr%C3%A9lation_(statistiques)) dans les statistiques des joueurs de la NBA pour la saison 2024-2025. Nous importerons les données, puis utiliserons Plotly Express pour afficher une représentation colorée de la [matrice](https://fr.wikipedia.org/wiki/Matrice_(math%C3%A9matiques)) de corrélation.

```python
import pandas as pd
import plotly.express as px
df = pd.read_csv('https://raw.githubusercontent.com/Data-Dunkers/data/refs/heads/main/NBA/player/nba_player_stats_2024-2025.csv')
px.imshow(df.corr(numeric_only=True),
          title='Matrice de corrélation des statistiques des joueurs de la NBA (2024-2025)',
          height=800, text_auto='.2f',
          labels={'x': 'Variable en abscisse', 'y': 'Variable en ordonnée', 'color': 'Corrélation'})
```

Survole les différentes cases pour voir les noms des variables et leur degré de corrélation.

## Description des colonnes

- **GP** — Matchs joués
- **MIN** — Minutes par match
- **PTS** — Points par match
- **FGM / FGA / FG%** — Tirs réussis, tirs tentés et pourcentage de réussite
- **3PM / 3PA / 3P%** — Tirs à trois points réussis, tentés et pourcentage de réussite
- **FTM / FTA / FT%** — Lancers francs réussis, tentés et pourcentage de réussite
- **REB** — Rebonds
- **AST** — Passes décisives par match
- **STL** — Interceptions par match
- **BLK** — Contres par match
- **TO** — Ballons perdus par match
- **DD2** — [Doubles-doubles](https://fr.wikipedia.org/wiki/Double-double) : matchs où le joueur atteint au moins 10 dans *deux* des statistiques suivantes : PTS, REB, AST, STL, BLK
- **TD3** — [Triples-doubles](https://fr.wikipedia.org/wiki/Triple-double) : matchs où le joueur atteint au moins 10 dans *trois* des statistiques suivantes : PTS, REB, AST, STL, BLK
