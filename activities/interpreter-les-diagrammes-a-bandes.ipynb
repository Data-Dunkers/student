# Diagrammes à bandes

Pour créer un diagramme à bandes, nous allons importer des données de la saison 2024-2025 de la NBA.

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/Data-Dunkers/data/refs/heads/main/NBA/player/nba_player_stats_2024-2025.csv')
df
```

Le code ci-dessous filtre les données pour comparer deux joueurs qui ont des moyennes de points semblables, mais des styles de jeu différents.

```python
player1 = "Pascal Siakam"
player2 = "Stephen Curry"
comparison = df[(df['Name'] == player1) | (df['Name'] == player2)]
comparison
```

Maintenant que nous avons filtré les données à comparer, nous pouvons créer un diagramme à bandes.

```python
import plotly.express as px

statistiques = {
    'FG%': 'Tirs réussis (%)',
    '3P%': 'Tirs à trois points réussis (%)',
    'PTS': 'Points par match',
    'AST': 'Passes décisives par match',
    'REB': 'Rebonds par match'
}
px.bar(comparison.rename(columns=statistiques), x='Name', y=list(statistiques.values()),
       barmode='group', labels={'Name': 'Nom', 'variable': 'Statistique', 'value': 'Valeur'},
       title=f'Comparaison de {player2} et de {player1}')
```
