# Leçon : Interpréter les nuages de points

Les nuages de points servent à représenter la relation entre deux variables. Nous utiliserons un jeu de données sur les statistiques des joueurs de la NBA pour la saison 2024-2025 afin de voir s’il existe une relation entre le nombre de **points** qu’un joueur marque et son **efficacité** au tir.

Nous comparons `PTS` (points par match) et `FG%` (pourcentage de tirs réussis).

La visualisation comprend aussi une **droite de tendance** (ou un **ajustement linéaire**) pour nous aider à voir la direction générale de la relation.

```python
import pandas as pd
import plotly.express as px

url = 'https://raw.githubusercontent.com/Data-Dunkers/data/refs/heads/main/NBA/player/nba_player_stats_2024-2025.csv'
df = pd.read_csv(url)

px.scatter(df, x='PTS', y='FG%', trendline='ols', hover_data=['Name'],
           labels={'PTS': 'Points par match', 'FG%': 'Pourcentage de tirs réussis', 'Name': 'Nom'},
           title='Pourcentage de tirs réussis et points par match dans la NBA (2024-2025)')
```
